```python
#imort pandas
import pandas as pd
```

```python
#load Instagram_Analytics.csv file
df = pd.read_csv("../Instagram_Analytics.csv")
```

```python
#print first five records
print("===== FIRST FIVE RECORDS =====")
df.head()
```

```python
#print shape of dataset
print("\n===== SHAPE OF DATASET =====")
print(df.shape)
```

```python
#print summary of data set 
print("\n===== SUMMARY =====")
display(df.describe(include="all"))
```

```python
#filter follower count having more than 1000 followers
print("\n===== POSTS WITH FOLLOWER COUNT > 10000 =====")
display(df[df["follower_count"] > 10000].head())
```

```python
#display account type who are creators 
print("\n===== CREATOR ACCOUNTS =====")
display(df[df["account_type"] == "creator"].head())
```

```python
#print accounts which has performance_bucket_label are viral 
print("\n===== VIRAL POSTS =====")
display(df[df["performance_bucket_label"] == "viral"].head())
```

```python
#sort by follower count 
print("\n===== SORT BY FOLLOWER COUNT (DESCENDING) =====")
display(df.sort_values(by="follower_count", ascending=False).head())
```

```python
#print follower count statistics
print("\n===== FOLLOWER COUNT STATISTICS =====")

print("Mean :", df["follower_count"].mean())

print("Median :", df["follower_count"].median())

print("Mode :")
print(df["follower_count"].mode())

print("Standard Deviation :", df["follower_count"].std())
```

```python
#print descriptive statistics 
print("\n===== DESCRIPTIVE STATISTICS =====")
display(df.describe())
```

```python
#average followers by account type 
print("\n===== AVERAGE FOLLOWERS BY ACCOUNT TYPE =====")

account_avg = df.groupby("account_type")["follower_count"].mean()

display(account_avg)
```

```python
#print correlation matrix 
print("\n===== CORRELATION MATRIX =====")

corr = df[
    [
        "follower_count",
        "comments",
        "shares",
        "saves",
        "reach",
        "impressions",
        "engagement_rate",
        "followers_gained",
        "caption_length",
        "hashtags_count"
    ]
].corr()

display(corr)
```
