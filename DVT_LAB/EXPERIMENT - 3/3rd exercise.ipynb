```python
# Import the required libraries
import pandas as pd
import numpy as np

# Create a sample DataFrame with missing values, duplicates, and inconsistent data
df = pd.DataFrame({
    "Name": [
        "Alice",
        "Bob",
        "Alice",      # Duplicate name
        "David",
        "Nyang",
        "Frank"
    ],
    "Age": [
        25.0,
        np.nan,       # Missing age
        25.0,
        17.0,
        30.9,
        999.0         # Invalid/outlier age
    ],
    "Salary": [
        50000.0,
        60000.0,
        50000.0,
        45000.0,
        np.nan,       # Missing salary
        70000.0
    ],
    "Place": [
        "Chennai",
        "Delhi",
        "Chennai",
        np.nan,       # Missing place
        "Mumbai",
        "Delhi"
    ],
    "Joined": [
        "2021",
        "2026",
        "2021",
        "2019",
        "2022",
        "2023"
    ]
})

# Display the DataFrame
print(df)
```

```python
print(df.dropna()) #TO DROP NULL COLUMNS
```

```python
print(df.dropna(how="all")) #TO DROP NULL COLUMNS WHEN ALL VALUES ARE NULL 
```

```python
print(df.dropna(thresh=4)) #TO DROP NULL ROWS IF THERE ARE MORE THAN 4 ROWS ARE NULL 
```

```python
print(df.dropna(subset=["Name", "Salary"])) # TO DROP ROWS IF THERE IS NULL IN NAME OR SALARY
```

```python
print(df.duplicated()) # TO PRINT THE BOOLEAN VALUES OF DUPLICATED ROWS 
```

```python
print(df.duplicated().sum()) #TO PRINT NUMBER OF ROWS OF DUPLICATEDE VALUES 
```

```python
# Display all duplicate rows in the DataFrame
print(df[df.duplicated(keep=False)])
```

```python
# Remove duplicate rows and store the cleaned DataFrame in a new variable
df_clean = df.drop_duplicates()

# Display the cleaned DataFrame
print(df_clean)
```

```python
# Display the number of rows and columns after removing duplicates
print(df_clean.shape)
```

```python
# Remove duplicate rows based on the 'Name' and 'Age' columns
# Keep the last occurrence of each duplicate and update the original DataFrame
df.drop_duplicates(
    subset=["Name", "Age"],
    keep="last",
    inplace=True
)

# Display the updated DataFrame
print(df)
```

```python
print(df.dtypes) #to print data types of columns 
```

```python
# Fill missing values in the 'Age' column with 0 and convert the data type to integer
df["Age"] = df["Age"].fillna(0).astype(int)

# Display the updated 'Age' column
print(df["Age"])

# Display the data types of all columns
print(df.dtypes)
```

```python
# Convert the 'Joined' column from string format to datetime format
df["Joined"] = pd.to_datetime(df["Joined"], format="%Y")

# Display the converted 'Joined' column
print(df["Joined"])

# Display the data type of the 'Joined' column
print(df["Joined"].dtype)
```

```python
# Convert the 'Age' column to numeric values and replace invalid entries with NaN
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")

print(df["Age"])
print(df.dtypes)
```

```python
#to rename columns
df.rename(
    columns={
        "Name": "Full_Name"
    },
    inplace=True
)

print(df)
```

```python
df.rename(
    columns={
        "Age": "Years",
        "Salary": "Income",
        "Place": "City",
        "Joined": "Joining_Date"
    },
    inplace=True
)

print(df)
```
