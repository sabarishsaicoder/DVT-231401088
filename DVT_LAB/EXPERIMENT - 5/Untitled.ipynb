```python
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt

# Load the Instagram Analytics dataset
df = pd.read_csv("../Instagram_Analytics.csv")

# Display the first five records
print("===== FIRST FIVE RECORDS =====")
display(df.head())
```

```python
# Check the shape of the dataset
print("===== SHAPE OF DATASET =====")
print(df.shape)
# Check the data types of all columns
print("===== DATA TYPES =====")
print(df.dtypes)
# Display descriptive statistics for numerical columns
print("===== DESCRIPTIVE STATISTICS =====")
display(df.describe())
```

```python
# Create a line chart for engagement rate over time

daily_engagement = df.groupby(
    "post_date"
)["engagement_rate"].mean()

plt.figure(figsize=(10,5))

plt.plot(
    daily_engagement.index,
    daily_engagement.values
)

plt.title("Instagram Engagement Rate Trend")
plt.xlabel("Post Date")
plt.ylabel("Average Engagement Rate")

plt.xticks(rotation=45)

plt.show()
```

```python
# Create a bar chart for average engagement rate by media type

media_engagement = df.groupby(
    "media_type"
)["engagement_rate"].mean()

plt.figure(figsize=(7,5))

plt.bar(
    media_engagement.index,
    media_engagement.values
)

plt.title("Average Engagement Rate by Media Type")
plt.xlabel("Media Type")
plt.ylabel("Average Engagement Rate")

plt.show()
```

```python
# Create a pie chart showing the distribution of media types

media_counts = df["media_type"].value_counts()

plt.figure(figsize=(7,7))

plt.pie(
    media_counts,
    labels=media_counts.index,
    autopct="%1.1f%%"
)

plt.title("Instagram Media Type Distribution")

plt.show()
```

```python
# Create a histogram for engagement rate distribution

plt.figure(figsize=(8,5))

plt.hist(
    df["engagement_rate"],
    bins=20
)

plt.title("Distribution of Engagement Rate")
plt.xlabel("Engagement Rate")
plt.ylabel("Frequency")

plt.show()
```

```python
# Create a scatter plot showing the relationship between followers and engagement rate

plt.figure(figsize=(8,5))

plt.scatter(
    df["follower_count"],
    df["engagement_rate"]
)

plt.title("Follower Count vs Engagement Rate")
plt.xlabel("Follower Count")
plt.ylabel("Engagement Rate")

plt.show()
```

```python
# Create a box plot to identify the distribution and outliers of engagement rate

plt.figure(figsize=(6,5))

plt.boxplot(
    df["engagement_rate"]
)

plt.title("Box Plot of Engagement Rate")
plt.ylabel("Engagement Rate")

plt.show()
```

```python
# Create multiple line charts comparing reach and impressions

daily_data = df.groupby("post_date")[
    ["reach", "impressions"]
].mean()

plt.figure(figsize=(10,5))

plt.plot(
    daily_data.index,
    daily_data["reach"],
    label="Reach"
)

plt.plot(
    daily_data.index,
    daily_data["impressions"],
    label="Impressions"
)

plt.title("Reach vs Impressions Over Time")
plt.xlabel("Post Date")
plt.ylabel("Average Count")

plt.legend()

plt.xticks(rotation=45)

plt.show()
```
